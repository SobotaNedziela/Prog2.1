```python
#Galgje 

#toon het word
def toon_woord(woord, geraden_letters):
    resultaten = ""
    for letter in woord:
        if letter in geraden_letters:
            resultaten += letter
        else:
            resultaten += "_"
    return resultaten




#letter_vragen
def letter_vragen(geraden_letters):
    letter = input("Raad een letter: ").lower()
    while letter in geraden_letters:
        letter = input("Deze letter is al geraden").lower()
    return letter
        
#is_spel_klaar?
def is_spel_klaar(woord, geraden_letters, fouten, max_fouten):
    if toon_woord(woord, geraden_letters) == woord:
        print("Gewonnen!")
        return True
    elif fouten == max_fouten:
        print("Verloren")
        return True
    else:
        return False

def galgje():
    woord = input("Vul een woord in: ").lower()
    geraden_letters = []
    fouten = 0
    max_fouten = 5
    
    while not is_spel_klaar(woord, geraden_letters, fouten, max_fouten):
        print(toon_woord(woord, geraden_letters))

        letter = letter_vragen(geraden_letters) 
    
        if letter in woord:
            geraden_letters.append(letter)
        else:
            fouten += 1 
            print("deze letter zit niet in het woord ")

galgje()
```
