```python
import random 
import math 
```

```python
#Aftand dobbelsteen berekenen

def dice(n):
    total = 0
    for i in range(n):
        total += random.randint(1,6) 
    return total

x_1 = dice(1)
y_1 = dice(1)

x_2 = dice(2)
y_2 = dice(2)

afstand = math.sqrt((x_1- x_2)**2 + (y_1 - y_2)**2)

print(f"Eerste coordinaat: ({x_1}, {y_1})")
print(f"Tweede coordinaat: ({x_2}, {y_2})")
print(f"De afstand is {afstand:.2f}")
```

```python
#Snelheid berekenen 

s = int(input("Vul het aantal seconden in"))

def tijd(s):
    uren = s//3600
    minuten = (s % 3600) // 60 
    seconden = (s % 60)
    print(f"{uren} uren, {minuten} minuten, {seconden} seconden")
    
tijd(s)
```

```python
#BMI berekenen 

g = float(input("Vul je gewicht in kg in "))
l = int(input("Vul je lengte in cm in "))

def bmi(g, l):
    bmi = g/((l/100)*(l/100))
    return bmi
    
print(f"Je BMI is {bmi(g, l):.2f}")
```

```python
#Rekenmachine maken 

x = int(input("Kies het getal x "))
y = int(input("Kies het getal y "))
opperator = input("Kies een opperator + , -, /, * ")

def rekenmachine(x, y, opperator):
    if opperator == "+":
        r = x + y 
    elif opperator == "-":
        r = x - y 
    elif opperator == "/":
        r = x/y
    elif opperator == "*":
        r = x*y
    return (r)

print(rekenmachine(x, y, opperator))
```
