```python
#Delers
import os
os.getcwd()

bestand = input("Vul een bestandsnaam in: ")

with open(bestand, "r") as file:
    regels = file.readlines()


#Van 15/n een 15 maken
s = int(regels[0])

#Een lege lijst getallen aanvullen met ieder element vanaf regel 3
getallen = []
for i in regels[2:]:
    x = int(i)
    getallen.append(x)
print(getallen)

#a en b zoeken in de lengte van de getallen
aantal = 0
for a in range(len(getallen)):
    for b in range(a+1, len(getallen) + 1):
        if sum(getallen[a:b]) == s:
            aantal += 1
            print(f"a en b{a, b}")

#print het aantal oplossingen         
if aantal == 0:
    print("Geen oplossingen")
else: 
    print(f"Aantal oplossingen {aantal}")
```
