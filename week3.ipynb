```python
#Cijferlijst analyse 

s = []    

for i in range(10):
    s.append(random.randrange(1, 11))
    
print(s)

#Print de onvoldoendes, voldoendes, en gemiddelde cijfer 
def voldoendes(s):
    o = 0
    v = 0
    for cijfers in s:
        if cijfers < 6:
            o += 1
        else:
            v += 1
    gemiddelde_s = sum(s)/len(s)
    return o, v, gemiddelde_s

voldoendes(s)

#Print een bericht naar de geslaage/gezakte en vertel
def geslaagd(s):
    o = 0
    v = 0
    for cijfers in s:
        if cijfers < 6:
            o += 1
        else:
            v += 1
    gemiddelde_s = sum(s)/len(s)
    if o <= 4 and gemiddelde_s >= 5.5:
        print(f"Je bent geslaagd met {v} voldoendes en {gemiddelde_s} gemiddeld ")
    else:
        print(f"Je bent gezakt met {o} onvoldoendes en {gemiddelde_s} gemiddeld")
    return o, v, gemiddelde_s

print(geslaagd(s))
```

```python
#Bioscoopzaal

b = []

#Maakt een bioscoopzaal met 15 stoelen
for i in range(15):
    if random.random() <= 0.4:
        b.append("#")
    else:
        b.append("L")  
print(b)
b_origineel = b.copy()

#5 personen komen de zaal binnen, opzoek naar een plek
for n in range(5):
    plek = random.randrange(0, 15)
    if b[plek] == "L":
         b[plek] = "#" 
    else:
        pass
    
print(b) 

#Vergelijkt de orspronkelijke bezetting met de nieuwe 
if b_origineel == b:
    print(" de bezetting is hetzelfde ")
else:
    print(" de bezetting is veranderd ")
```
