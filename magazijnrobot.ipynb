```python
 #Gestoorde magazijnrobot
    
vc = "VvaAlLVAAVLR"


alfabet = "abcdefghijklmnopqrstuvwxyz"

oude_lengte = 0

while len(vc) != oude_lengte:
    oude_lengte = len(vc)
    
    for letter in alfabet:
        c_1 = letter + letter.upper()
        c_2 = letter.upper() + letter
        
        vc = vc.replace(c_1, "")
        vc = vc.replace(c_2, "")
        
print(vc)

rijen = list(range(0, 16))
kolommen = list(range(0, 4))


for letter in vc[:4]:
    midden = len(rijen) // 2 
    
    if letter == "A":
        rijen = rijen[midden:]
    elif letter == "V":
        rijen = rijen[:midden]
    
rij = rijen[0]
    
for letter in vc[4:]:
    midden = len(kolommen) // 2
    
    if letter == "R":
        kolommen = kolommen[midden:]
    elif letter == "L":
        kolommen = kolommen[:midden]
        
kolom = kolommen[0]

locatie_id = rij * 4 + kolom
        

print(locatie_id)
```
