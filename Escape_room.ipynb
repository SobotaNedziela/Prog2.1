```python
import random
import math

team_namen = ["Alpha", "Puzzelaars", "Alpha", "Escape Kings", "Puzzelaars", "De Koplopers"]
gespeelde_tijden = [55, 62, 48, 70, 59, 45]


#Unieke teams bepalen 
def bepaal_unieke_teams(team_namen):
    teams = set()
    for naam in team_namen:
        teams.add(naam)
    return teams

#Snelste tijd per team
def registreer_record_tijden(namen, tijden):
    d = {}
    for naam, tijd in zip(namen, tijden):
        if naam not in d:
            d[naam] = tijd
        else:
            if tijd < d[naam]:
                d[naam] = tijd
    return d
            
#Ontsnapt of niet
def is_ontsnapt(tijd):
    if tijd <= 60:
        return True
    else:
        return False
    
def filter_geslaagd(record_dict):
    geslaagden = [(team, tijd) for team, tijd in record_dict.items() 
                if is_ontsnapt(tijd)]
    return geslaagden

#Dobbelen voor korting
def bepaal_korting(seed_waarde):
    random.seed(seed_waarde)
    korting = 10
    worp = 0
    while worp != 6:
        worp = random.randint(1, 6)
        if worp != 6:
            korting -= 2
    return korting

uniek = bepaal_unieke_teams(team_namen)
print("Unieke teams:", uniek)

records = registreer_record_tijden(team_namen, gespeelde_tijden)
print("Recordtijden:", records)

geslaagd = filter_geslaagd(records)
print("Geslaagde teams:", geslaagd)

korting_gewonnen = bepaal_korting(2026)
print(f"Het team {geslaagd[0][0]} wint een korting van €{korting_gewonnen},-!")
```
