```python
#Sjoelen voor 40 punten...

#Eerst een fucntie om de punten te tellen
def punten_tellen(sjoelbak):
    
    complete_set = min(sjoelbak)
    
    totaal_punten = complete_set * 20
    
    totaal_punten += (sjoelbak[0] - complete_set) * 1
    totaal_punten += (sjoelbak[1] - complete_set) * 2
    totaal_punten += (sjoelbak[2] - complete_set) * 3    
    totaal_punten += (sjoelbak[3] - complete_set) * 4
    
    return totaal_punten

test_sjoelbak = [4, 6, 4, 5] 
print(punten_tellen(test_sjoelbak))
```

```python
#Deze functie simuleert één beurt 
import random

def beurt(sjoelbak, schijven_over):
    
    while schijven_over > 0:
        schijf_raak = random.random()
        schijven_over -= 1
        aantal_raak = 0
        
        if schijf_raak < 0.333:
            gekozen_vak = random.randint(0, 3)
            sjoelbak[gekozen_vak] += 1
            aantal_raak += 1
            
    
    return aantal_raak, sjoelbak

print(beurt(test_sjoelbak, 20))
```

```python
#De stand uitprinten na een beurt 

def print_stand(beurt_nummer, aantal_raak, sjoelbak):
    print(f"Stand na beurt {beurt_nummer}")
    print(f"Er waren {aantal_raak}")
    print(f"Vak 1: {sjoelbak[0]} schijven")
    print(f"Vak 2: {sjoelbak[1]} schijven")
    print(f"Vak 3: {sjoelbak[2]} schijven")
    print(f"Vak 4: {sjoelbak[3]} schijven")
    print(f"Dit levert {punten_tellen(sjoelbak)} punten op") 
```

```python
#Simulatie 

sjoelbak = [0, 0, 0, 0]

for beurt_nummer in range(1, 4):
    print(f"Beurt {beurt_nummer}:")
    
    schijven_over = 30 - sum(sjoelbak)
    print(f"Er zijn nog {schijven_over} om mee te sjoelen")
    
    aantal_raak, sjoelbak = beurt(sjoelbak, schijven_over)
    
    print_stand(beurt_nummer, aantal_raak, sjoelbak)
    
```
