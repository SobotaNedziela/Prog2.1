```python
### Hoger lager

import math
import random

#pak een random getal tussen [1, 101)
x = random.randrange(1, 101)


#vraagt om een input i, zolang i niet gelijk is aan x vraagt de speler om nog een keer.
#geeft ook aan de speler aan of zijn getal hoger of lager is dan x
def raad_getal(x):
    i = int(input("Raad het getal x tussen 1 en 100: "))
    while i != x:
        i = int(input("Fout! Raad het nog een keer: "))
        if i > x:
            print("Het getal is lager")
        elif i < x:
            print("Het getal is hoger")
    print("Je hebt het goed!!!")
    return i

raad_getal(x)
```

```python
#Gemiddelde berekenen 
import math

def gemiddelde_calc():
    getallen = []

    x = input("Vul een getal x in of klaar: ")
    
    while x != "klaar":
        getallen.append(int(x))
        x = input("Vul nog een getal x in of klaar: ")
    
    gemiddelde = sum(getallen) / len(getallen)
    print(f"Het gemiddelde van x is {gemiddelde}")
    return gemiddelde
        
print(gemiddelde_calc())
```

```python
#Aftelklok 

import time

#We willen de fucntie aanhebben zolang s niet gelijk is aan nul
#Dan telt s af met s-1 elke seconde
#Klaar met tellen? -> vertel dat
def aftellen():
    s = int(input("Vul een geheel getal in: "))
    
    while s != 0:
        time.sleep(1)
        s = s - 1
        print(s)
    print("afgeteld tot nul!")

aftellen()
```
