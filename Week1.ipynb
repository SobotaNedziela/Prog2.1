```python
#BMI berekenen 

gewicht = float(input("Vul je gewicht in kilo's in "))
lengte = float(input("Vul je lengte in centimers in "))

print(f"Jouw BMI is {gewicht/(lengte/100)}")

#{}/{} -> tekst delen door tekst, kant niet. 
#Wat wel kan is {a/b} -> a/b en dan er tekst van maken met fstring
```

```python
#Opbrengst spaarrekening berekenen 

s = float(input("Vul u spaargeld in tot de cent "))
r = float(input("Vul u rente percentage per jaar in als komma getal "))
t = float(input("Vul het aantal jaar in dat het geld blijft liggen "))

w = s*(1+r)**t

print(f"Het bedrag van {s} euro is over {t} jaar {w:.2f} euro waard ")

#:.2f zorgt ervoor dat er iets op 2 getallen na de komma wordt afgerond en fixed is
```

```python
#van Celcius naar Fahrenheit 

temp = int(input("Vul de temperatuur in Celcius in "))

temp_f = float(temp * 1.6 + 32)

print(f"Jouw temperatuur in Fahrenheit is {temp_f}")
```

```python
#korting berekenen

p = float(input("Vul de prijs in "))

k = float(input("Vul de korting in als komma getal "))

t = input("Vul in of de korting wordt toegepast 1 = Ja") == "1"

e_p = p*(1 - (k*t))

print(f"De prijs van jouw product is {e_p:.2f}")
```
