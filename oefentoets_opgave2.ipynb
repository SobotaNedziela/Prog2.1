```python
#Binary boarding opgave 2 

def wat_is_stoel_id(boarding_pass):
    rijen = list(range(0, 128))
    kolommen = list(range(0, 8))

    for letter in boarding_pass[:7]:
        midden = len(rijen) // 2
    
        if letter == "F":
            rijen = rijen[:midden]
        elif letter == "B":
            rijen = rijen[midden:]
        
    rij = rijen[0]
    
    for letter in boarding_pass[7:]:
        midden = len(kolommen) // 2
        
        if letter == "L":
            kolommen = kolommen[:midden]
        elif letter == "R":
            kolommen = kolommen[midden:]
        
    kolom = kolommen[0]
    
    id_stoel = rij * 8 + kolom   

    return rij, kolom, id_stoel

file = open("input_opgave_2.txt")
alle_ids = []

for regel in file:
    
    pas = regel.strip()
    stoel = wat_is_stoel_id(pas)
    alle_ids.append(stoel)
    
file.close()

print(max(alle_ids))
```
